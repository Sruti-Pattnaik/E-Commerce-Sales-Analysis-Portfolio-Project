OUTLIERS

Detect Outliers in Multiple Columns

```python
columns = ["Sales", "Profit", "Quantity", "Discount"]

for col in columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = df[(df[col] < lower) | (df[col] > upper)].shape[0]

    print(f"{col}: {count} outliers")
```
